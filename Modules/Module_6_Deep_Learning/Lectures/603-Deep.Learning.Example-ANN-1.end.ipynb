## Linear regression ... rewritten


Secondary school notation.  It's limited in its notation to handle more than one factor, i.e. x.

$ y = mx + b $

---

Rewritten using $ \beta $ and rearranging terms.

$ y = \beta_1 x + b $

$ y = \beta_1 x + \beta_0 $

$ y = \beta_0 + \beta_1 x $

---

Rewritten using subscripts on factors

$ y = \beta_0 x_0 + \beta_1 x_1  ; x_0 = 1 $

We now have everything written using pairs of a $ \beta $ and a factor, both having subscripts.

For example: $ \beta_1 x_1 $

---

Rewritten using more than one factor.  That is, we can add more complexity to the formula by adding a new $ \beta $ and factor pair.

$ y = \beta_0 x_0 + \beta_1 x_1 + \beta_2 x_2 + ... + \beta_n x_n $

---

Rewritten as vectors.  That is, let's split the $ \beta $s from the factors, each into their own list, i.e. array or vector.  The requirement is that each $ \beta $ is paired with a factor.  That is, the lengths ( or dimensions, sizes ) of both vectors are the same.

$ \mathbf{B} = [\beta_0, \beta_1, \beta_2, ..., \beta_n ] $

$ \mathbf{X} = [x_0, x_1, x_2, ..., x_n ] ; x_0 = 1 $

$ ( \mathbf{X} {\odot} \mathbf{B} ) = ( \beta_0 x_0, \beta_1 x_1, \beta_2 x_2, ..., \beta_n x_n ) ; x_0 = 1 $

Condensing the notation using summation, i.e. $ \Sigma $.


$ {\Sigma} ( \mathbf{X} {\odot} \mathbf{B} ) = \beta_0 x_0 + \beta_1 x_1 + \beta_2 x_2 + ... + \beta_n x_n ; x_0 = 1 $

---

Rewritten to further condense the notation into a Dot product.

$ ( \mathbf{X} \cdot \mathbf{B} ) $

---
Rewritten in numpy

```python
B = np.array( [ 1, 2, 3, 4, 5] )
X = np.array( [ 1, 2, 2, 2, 2] )
(X * B).sum()
X @ B
```

## A single perceptron


```python
# A neural network with 5 nodes in the input layer, and 1 node in the output layer, plus weights and bias
Inputs = np.array( [ -30, 1, 2, 3, 4, 5, ])
Weights = np.array( [ 1, 2, 2, 2, 2, 2] )

(Inputs * Weights), (Inputs * Weights).sum(), Inputs @ Weights
```

```python
# Apply a threshold
((Inputs @ Weights) > 0) + 0
```
