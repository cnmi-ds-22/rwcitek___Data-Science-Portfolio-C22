# GitHub Organizations

## Introduction: Organizing Your Code with GitHub Organizations

As you build more projects, your personal GitHub profile can quickly become a crowded, endless list of repositories. Finding a specific project from last month, separating your course exercises from your personal portfolio, or keeping raw datasets apart from production code can become a real headache.

To solve this, professionals use **GitHub Organizations** as a powerful tool for visual and structural organization.

Think of a GitHub Organization not as a "company," but simply as a **dedicated, isolated folder** for a specific collection of work.

* It gives you a brand-new, completely separate dashboard out of your main profile's way.
* It allows you to group related repositories together under a single, meaningful umbrella name.

We are going to use Organizations to cleanly separate our workspace. You are going to create a brand-new Organization to act as a dedicated folder.

Then you are going to **Fork** (make a clean copy of) repositories from your cohort into that new workspace.



## Technical Setup: Creating an Organized Workspace

To get started, we need to create a centralized home for your code and copy your corhort's starting repositories into it. Follow these two steps:


### Step 1: Create the GitHub Organization

GitHub organizations allow multiple people to collaborate on repositories simultaneously.

1. Log into your personal account on [GitHub](https://github.com).
1. Click your profile picture in the top-right corner and select **Organizations**.
1. Click the **New organization** button.
1. Choose the **Free** tier (this is perfectly sufficient for our project work).
1. Fill out the details:
    * **Organization name:** Enter `ddds-c{cohort_number}-{your_username}-test1`
    * **Contact email:** Your primary email.
    * **Account ownership:** Select "My personal account."

1. *(Optional)* **Add organization members**: you can click **Skip this step** at the bottom to jump straight to your new dashboard.



### Step 1b: Delete the organization
We are going to delete the organization so that you are familiar with that process.

1. Log into your personal account on [GitHub](https://github.com).
1. Click your profile picture in the top-right corner and select **Your organizations**.
1. Scroll down the page until you find your organization's name, `ddds_c{cohort_number}_{your_username}_test1`,  and click on it.
1. Click on **Settings** in the top-right corner
1. Scroll down to the ***Danger Zone*** and click on the **Delete this organization** button
1. Enter this organization's name and click **Cancel plan and delete the organization**.




### Step 1c:

Create a new organization using the instructions from Step 1 above, naming it `ddds-c{cohort_number}-{your_username}`.




### Step 2: Fork the Project Repositories into the Org

Now you need to pull your cohort's code into your new organization.

1. Navigate to the GitHub URL of the original repository.
2. In the top-right corner of the page, click the **Fork** button.
3. Under the **Owner** dropdown menu, **do not** select your personal username. Instead, select your newly created **GitHub Organization**.
4. Leave the repository name as-is.
5. Click the green **Create fork** button.

GitHub will take a few seconds to copy the repository. Once it's done, you'll see the repository live under your organization's URL ( e.g., github.com/your-org-name/repo-name ), ready for you to use.
