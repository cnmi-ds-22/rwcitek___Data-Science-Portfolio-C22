# Example Cloud Functions


**DigitalOcean Functions** is a serverless computing solution that allows developers to run event-driven code on demand without provisioning, managing, or paying for idle servers. With native support for Python and automatic scaling, it is particularly useful in data science for deploying lightweight machine learning models as low-latency inference APIs. Data scientists can also leverage these functions to automate scheduled data ingestion, preprocessing pipelines, and webhook-triggered ETL tasks while keeping infrastructure overhead minimal. Because compute is billed strictly to the millisecond of execution, it provides a highly cost-effective way to handle sporadic workloads and real-time predictions without maintaining dedicated virtual machines.


- [Cloud function pricing]( https://www.digitalocean.com/pricing/functions )


## Setup


### Create a Digital Ocean API token



1. Visit API tokens:
    - https://cloud.digitalocean.com/account/api/tokens
1. Give it a name, e.g. function token
1. Choose expiration date, e.g. 30 days
1. Select Custome scopes
1. Select function scope: all
1. Click Generate Token
1. Accept any suggested scopes
1. Copy the token to a password manager
1. Copy the token to your Colab Secret vault using the name "do_token"

### Install doctl


```python
%%bash

version=$(
  curl -fsSIL -o /dev/null -w '%{url_effective}\n' https://github.com/digitalocean/doctl/releases/latest |
    sed -E 's#.*/v?##'
)

curl -sL https://github.com/digitalocean/doctl/releases/download/v${version}/doctl-${version}-linux-amd64.tar.gz |
  tar -xz -C /usr/local/bin

which doctl
doctl version
```
