## doctl


see
- https://gist.github.com/rwcitek/1a03bfd7b59595f4e1e36f8dc22cf393#file-digital-ocean-function-docker-md


```python
%%bash

version=$(
  curl -fsSIL -o /dev/null -w '%{url_effective}\n' https://github.com/digitalocean/doctl/releases/latest |
    sed -E 's#.*/v?##'
)

curl -sL https://github.com/digitalocean/doctl/releases/download/v${version}/doctl-${version}-linux-amd64.tar.gz |
  tar -xz -C /usr/local/bin

which doctl
doctl version
```
