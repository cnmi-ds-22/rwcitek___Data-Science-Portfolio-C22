```python
%%bash

# installing charliecloud

cat <<'eof' > /tmp/charlie-cloud.sh
  cd /content/
  apt-get update && \
    apt-get install -y build-essential python3 python3-pip git wget squashfs-tools
  git clone https://github.com/hpc/charliecloud.git
  cd charliecloud

  ./autogen.sh
  ./configure --prefix=$HOME/.local
  make
  make install

  echo == Done
  sleep 3000
  rm -f "${0}"
eof

tmux new -s charlie-cloud -d "bash /tmp/charlie-cloud.sh"
```

```python

%%markdown

# using charliecloud

export CH_IMAGE_STORAGE=$HOME/.charliecloud
export USER=root

ch-image pull alpine:latest
ch-run alpine:latest -- cat /etc/os-release
ch-run alpine:latest -- /bin/sh
ch-run alpine:latest -- sleep infinity & BG_PID=$! ; echo $BG_PID
ch-run --join-pid=$BG_PID alpine:latest -- /bin/sh


ch-image pull ubuntu:26.04
ch-run ubuntu:26.04 -- cat /etc/os-release
ch-run ubuntu:26.04 -- /bin/bash
ch-run -W -t ubuntu:26.04 -- sleep infinity & BG_PID=$! ; echo $BG_PID
ch-run --join-pid=$BG_PID ubuntu:26.04 -- /bin/bash

apt-get update && apt-get install -y curl

```
